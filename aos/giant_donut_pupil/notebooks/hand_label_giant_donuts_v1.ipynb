## Description

Hand labels for the v2 giant-donut selection (`study.md`, "selection plan v2", step C).
Twelve pages: rafts R22, R31, R13, R10, R40 and R04 on visits 2025102300337 (extra-focal)
and 2025102300340 (intra-focal). Each page is the background-subtracted raft from the 4x4
binned ISR cache, with every the_monster star at 7-14.5 AB mag (marker by magnitude bin) and
red crosses for peaks no catalog star explains.

How to label:
1. **Click a marker** to cycle its label: good -> marginal -> blended -> not_donut -> unlabelled.
2. **Click a donut with no marker** to add it as a missed donut (dashed circle, `good`); click
   it again to cycle its label.
3. Label **every donut on the page**, then tick "page done". Unlabelled markers on a done page
   count as not worth scoring.
4. Use the toolbar zoom/pan freely; clicks are ignored while a toolbar mode is active.

Labels: **good** = isolated, whole ring clean, usable for a pupil fit; **marginal** = usable,
but a neighbour's halo, an edge or a faint overlap touches it; **blended** = overlapped, not
usable; **not_donut** = nothing there.

**Output:** `hand_labels.csv` and `hand_labels_pages.json` in the `selection_diag` run directory,
rewritten on every click, so the session can stop and resume.

**Based on:** `code/hand_label.py`, `code/selection_v2.py`
