## Description

Hand review of the giant-donut selection (`study.md`, step C, reworked). Twelve pages: rafts
R22, R31, R13, R10, R40 and R04 on visits 2025102300337 (extra-focal) and 2025102300340
(intra-focal). Each page is the background-subtracted raft from the 4x4 binned ISR cache with
the CCD edges and two markers:

- **^**: a catalog star the current selection classes isolated;
- **v**: a good peak: a threshold_rel 0.3 detection, whole donut on silicon, not saturated,
  annulus_ratio < 0.05 and overlap_frac < 0.02 against every other peak and every relevant
  catalog neighbour.

A donut with both gets both, offset vertically. Marker colour: yellow unreviewed, green good,
red bad. The method is in `method.md`.

How to review:
1. **Click a bad marked donut** to mark it bad (red); click again to undo.
2. **Click a good donut with no marker** to add it as missed (dashed cyan circle); click it
   again to remove it.
3. Press **Done: the rest are good**. Every unreviewed marker on the page becomes good.
   On a done page, clicking a bad donut sets it back to good.
4. Use the toolbar zoom/pan freely; clicks are ignored while a toolbar mode is active.

**Output:** `hand_labels_v2.csv` and `hand_labels_v2_pages.json` in the `selection_diag` run
directory, rewritten on every click, so the session can stop and resume. `hand_labels.csv`
(v1) is not touched.

**Based on:** `code/hand_label.py`, `code/peak_quality.py`, `code/selection_v2.py`
